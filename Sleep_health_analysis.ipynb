```python
# Sleep data of a person

sleep_hours = 6.5
rem_hours = 1.2
core_hours = 3.8
deep_hours = 0.7
awake_minutes = 45
```

```python
# Calculate sleep score

sleep_score = 0

# Sleep duration
if 7 <= sleep_hours <= 9:
    sleep_score += 40
elif 6 <= sleep_hours < 7 or 9 < sleep_hours <= 10:
    sleep_score += 30
else:
    sleep_score += 20

# REM sleep
if rem_hours >= 1.5:
    sleep_score += 20
else:
    sleep_score += 10

# Core sleep
if core_hours >= 3.5:
    sleep_score += 20
else:
    sleep_score += 10

# Deep sleep
if deep_hours >= 1:
    sleep_score += 20
else:
    sleep_score += 10

print("Sleep Score:", sleep_score, "/ 100")
```

```python
# Find mental health category

if sleep_score >= 80:
    mental_health = "Good"
elif sleep_score >= 60:
    mental_health = "Moderate"
else:
    mental_health = "Needs Attention"

print("Mental Health:", mental_health)
```
